```python
import optuna
import pandas as pd

study = optuna.load_study(
    study_name="earthquake_architecture_search",
    storage="sqlite:///../architechturesearch/data/architecture_search.db"
)

df = study.trials_dataframe()

print(df.head())
```

```python
complete_df = df[df["state"] == "COMPLETE"]
```

```python
complete_df.sort_values("value")
```

```python
pruned_df = df[df["state"] == "PRUNED"]
```

```python
pruned_df.sort_values("value")
```

Based on these results, the clear ideas are that 
1. BatchNorm = False
2. Layers best options between 5 and 6
3. Initial filters best at 16 or 32
4. Kernel size 3 or 5 best 
5. fc size 128 or 256 initiall ybest

Bigger models does not suggest better results, run to run variance is high 

```python
study2= optuna.load_study(
    study_name="earthquake_architecture_search",
    storage="sqlite:///../architechturesearch/data/architecture_search2.db"
)

df2 = study2.trials_dataframe()

print(df2.head())
```

```python
completed2 = df2[df2["state"] == "COMPLETE"] 
```

```python
completed2.sort_values("value")
```

```python
meanof256 = completed2[completed2["params_fc_size"] == 256]["value"].mean()

meanof512 = completed2[completed2["params_fc_size"] == 512]["value"].mean()

print(meanof256)
print(meanof512)
```

```python
meanof256 = df2[df2["params_fc_size"] == 256]["value"].mean()

meanof512 = df2[df2["params_fc_size"] == 512]["value"].mean()

print(meanof256)
print(meanof512)
```
